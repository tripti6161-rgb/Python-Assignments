```python
#Bank Transactions:
# Positive values represent deposits. Find all Deposits and withdrawls also.
transactions = [5000, -2000, 3000, -1000, 7000, -500]
deposits = [ex for ex in transactions if ex>0]
print(deposits)
withdrawls = [ex for ex in transactions if ex<0]
print(withdrawls)
```

```python
#Give a 10% discount to products costing 1000  or more, otherwise give 5% and store result into new list
prices = [500, 1200, 800, 2000, 1500]
result =[ex*0.90 if ex>1000 else ex*0.95 for ex in prices]
print(result)
```

```python
# Find employee earning more than 50000.
employees = [
        {"name": "Amit", "salary": 45000},
        {"name": "Rahul", "salary": 60000},
        {"name": "Rohit", "salary": 35000},
        {"name": "Neha", "salary": 75000}
]
empNames=[x["name"] for x in employees if x['salary']>50000]
print(empNames)
```

```python
# Find names of employees who earne more than 60000 and belong to IT department.
employees = [
        {"name": "Amit", "department": "Sales", "salary": 55000},
        {"name": "Rahul", "department": "HR", "salary": 75000},
        {"name": "Rohit", "department": "IT", "salary": 85000},
        {"name": "Neha", "department": "IT", "salary": 70000}
]

empNames = [n['name'] for n in employees if n['salary']>60000 and  n['department'] == 'IT']
print(empNames)
```
